 
 
### Fundamental truth

A computer screen is **not a canvas**.
It is a **grid of tiny squares called pixels**.

Each pixel can:

* Be turned ON or OFF
* Or be given a color (RGB values)

So the *only thing* a computer can do is:

> 💡 **Decide which pixels to color, and with what values**

That’s it.

Everything else—lines, circles, images, text—is just **clever pixel selection**.

---

## 2️⃣ What Is Image Drawing?

### Definition (from first principles)

**Image drawing** means:

> Converting **mathematical shapes or images** into **colored pixels on a grid**

This conversion process is called:

> **Scan Conversion** or **Rasterization**

---
 

### 3.1 Raster Drawing (Pixel-Based)

* Draw directly on pixels
* Decide color pixel by pixel

Example:

```
Set pixel (x=10, y=20) to red
```

Used in:

* Screens
* Images
* Photos

📌 **Most basic and most important**

---

### 3.2 Vector Drawing (Math-Based)

* Draw shapes using equations
* Converted to pixels later

Example:

```
Line from (x1, y1) to (x2, y2)
Circle with center (x, y) and radius r
```

Used in:

* CAD
* SVG graphics
* Fonts

💡 **Vectors are ideas, raster is reality**

---

## 4️⃣ Drawing a Point (The Simplest Case)

### How to draw a point?

```
Choose pixel (x, y)
Assign color
```

That’s it.

> 💡 First principle:
> **Every complex drawing starts with a single pixel**

---
 

### Problem:

A line is continuous, but pixels are discrete.

So we must answer:

> Which pixels best represent a straight line?

---

### 5.1 Naive Idea (Why It Fails)

Use equation:

```
y = mx + c
```

For each x, compute y.

❌ Problems:

* Floating-point math
* Gaps in pixels
* Slow

---

### 5.2 Smart Idea: Incremental Drawing

Instead of perfect math:

* Move step by step
* Choose the *nearest pixel*

This leads to:

> **Bresenham’s Line Drawing Algorithm**

💡 First principle:

> **Choose the pixel that keeps you closest to the true line**

No floats. Only integers. Fast and accurate.

---

## 6️⃣ Drawing a Circle (Even More Interesting)

### Circle equation:

```
x² + y² = r²
```

Again, infinite points vs finite pixels.

### Key observation:

A circle is **symmetric**.

So:

* Compute one section
* Mirror it to others

This leads to:

> **Midpoint Circle Algorithm**

💡 First principle:

> **Exploit symmetry to reduce computation**

---

## 7️⃣ Drawing Shapes (Polygons)

### How polygons are drawn:

1. Draw lines between vertices
2. Close the shape

Triangle:

```
3 lines
```

Rectangle:

```
4 lines
```

Complex shapes:

```
Many connected lines
```

---

 
Drawing boundaries is not enough.
We often want **solid shapes**.

### Scan-Line Fill Technique

Idea:

1. Move row by row (scan line)
2. Find boundary intersections
3. Fill pixels in between

💡 First principle:

> **Inside a shape = between two boundaries**

---

## 9️⃣ Drawing Images (Bitmaps)

An image file already contains pixel values.

Drawing an image means:

```
For each pixel in image:
  Copy color to screen pixel
```

Operations include:

* Scaling
* Rotation
* Cropping
* Translation

All are just **pixel remapping**.

---

 

Transformations answer:

> Where should this pixel go?

### Common transformations:

* Translation (move)
* Scaling (resize)
* Rotation
* Reflection

All done using **matrix multiplication**.

💡 First principle:

> **Transform points first, draw later**

---

````python
import cv2
import matplotlib.pyplot as plt

camera = cv2.VideoCapture(0)

frame_width = int(camera.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(camera.get(cv2.CAP_PROP_FRAME_HEIGHT))

codex = cv2.VideoWriter_fourcc(*'XVID')
recorded = cv2.VideoWriter('recorded.avi', codex, 20.0, (frame_width, frame_height))

frame_count = 0

while True:
  success, image = camera.read()
  
  if not success:
    break
  recorded.write(image)
  
  # Display every 10th frame to avoid slowdown
  if frame_count % 10 == 0:
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.axis('off')
    plt.title(f'Frame {frame_count}')
    plt.tight_layout()
    plt.show()
  
  frame_count += 1
  
  # Press 'q' in the console or interrupt kernel to stop
  if frame_count > 100:  # Limit frames for testing
    break
    
camera.release()
recorded.release()
print(f"Video saved! Total frames: {frame_count}")
````

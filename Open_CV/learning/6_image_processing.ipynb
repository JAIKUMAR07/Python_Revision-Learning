## bit wise operation 

- iske through 2 images ko combine kr skte 
- cur out kr skte kuch chize 
- ye operation bit wise pixels me hota hai 
- and  - face area ko hi rkho 
- or - meger krna object ko 
- un wanted area ko remove 

```python
import cv2
import numpy as np

img1 = np.zeros((300, 300), dtype="uint8")
img2 = np.zeros((300, 300), dtype="uint8")
cv2.circle(img1, (150, 150), 100, 255, -1)
cv2.rectangle(img2, (50, 50), (250, 250), 255, -1)

bitwise_and = cv2.bitwise_and(img1, img2)
bitwise_or = cv2.bitwise_or(img1, img2)
bitwise_xor = cv2.bitwise_xor(img1, img2)
bitwise_not = cv2.bitwise_not(img1)

fig, axes = plt.subplots(2, 3, figsize=(12, 8))

axes[0, 0].imshow(img1, cmap='gray')
axes[0, 0].set_title("Image 1")
axes[0, 0].axis('off')

axes[0, 1].imshow(img2, cmap='gray')
axes[0, 1].set_title("Image 2")
axes[0, 1].axis('off')

axes[0, 2].imshow(bitwise_and, cmap='gray')
axes[0, 2].set_title("Bitwise AND")
axes[0, 2].axis('off')

axes[1, 0].imshow(bitwise_or, cmap='gray')
axes[1, 0].set_title("Bitwise OR")
axes[1, 0].axis('off')

axes[1, 1].imshow(bitwise_xor, cmap='gray')
axes[1, 1].set_title("Bitwise XOR")
axes[1, 1].axis('off')

axes[1, 2].imshow(bitwise_not, cmap='gray')
axes[1, 2].set_title("Bitwise NOT")
axes[1, 2].axis('off')

plt.tight_layout()
plt.show()
```
